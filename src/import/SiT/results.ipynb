```python
from pathlib import Path
import re

save_dir = Path('SiT/image_283_231_250_25')
png_files = list(sorted(save_dir.glob("*.png")))

paths = []
for path in png_files:
    path = str(path)
    name = path[path.rfind('/') + 1:]
    steps = int(name.split('.')[0][:])
    paths.append((steps, path))

paths = sorted(paths)
paths[:2]
```

```python
for steps, path in reversed(paths):
    print(250 - steps, 'cat', steps, 'dog')
    display(Image.open(path))
    print('=' * 50)
```
